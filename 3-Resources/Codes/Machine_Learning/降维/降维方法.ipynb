高维数据集非常稀疏，训练数据彼此之间相差很远，这样容易导致预测变得不可靠
**训练集的维度越高，过拟合的风险越大**

数据所需的训练数量随着维度的增加指数上升

高维度的特征会使训练缓慢，同时还会导致找到解决方案更加困难

降维的好处：
- 加速模型的训练（但同时会轻微降低系统性能）
- 可视化分析：将数据降维后进行绘制，通过视觉的观察，可以获得一些重要的洞察

**线性投影法**

假设：高维数据分布在线性子空间，即内在结构是线性平面 / 超平面 
使用场景：
1. 数据内在结构近似线性，特征之间线性相关
2. 需要可解释性、特征权重、方差分析
3. 降维结果需要用于后面的模型训练

**非线性投影法**
使用场景：
1. 数据非线性，但不算极端扭曲
2. 降维结果需要用于后面的模型训练

**流形学习**
假设：数据位于低维光滑流形上，即内在结构是卷曲、弯折、非线性曲面
