```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree

np.random.seed(42)

X_quad = np.random.rand(200, 1) - 0.5  
y_quad = X_quad ** 2 + 0.025 * np.random.randn(200, 1)

tree_reg = DecisionTreeRegressor(min_samples_leaf=10, random_state=42)

tree_reg.fit(X_quad, y_quad)
```

```python
plt.figure(figsize=(10,6))
# 生成密集采样点，用来画出连续预测线
X_smooth = np.linspace(-0.5, 0.5, 500).reshape(-1,1)
y_pred = tree_reg.predict(X_smooth)

# 原始数据点
plt.scatter(X_quad, y_quad, color="#556688", alpha=0.65, label="Raw Data")
# 决策树预测分段水平线
plt.plot(X_smooth, y_pred, color="#c0392b", lw=2.5, label="Decision Tree Prediction")

plt.xlabel("$x$", fontsize=12)
plt.ylabel("$y$", fontsize=12)
plt.title(f"DecisionTreeRegressor", fontsize=13)
plt.legend(fontsize=11)
plt.grid(alpha=0.22, linestyle="--")
plt.show()

# ===================== 4. 绘制回归树结构 =====================
# plt.figure(figsize=(12,6))
# plot_tree(tree_reg, rounded=True, filled=True, fontsize=10)
# plt.title("Regression Tree Structure")
# plt.show()
```
